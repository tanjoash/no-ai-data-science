```python
import pandas as pd
```

```python
adult_census = pd.read_csv("./dataset/adult-census.csv")
adult_census.head()
```

```python
target_class = "class"
target = adult_census[target_class]
target.head()
```

```python
data = adult_census.drop(columns=[target_class])
data.head()
```

```python
feature_columns = ["age", "capital-gain", "capital-loss", "hours-per-week"]
```

```python
data = data[feature_columns]
data.head()
```

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier()
_ = model.fit(data, target)
```

```python
target_predicted = model.predict(data)
```

```python
target_predicted
```

```python
print(
    "Number of correct prediction: "
    f"{(target == target_predicted).sum()} / {adult_census.shape[0]}"
)
print(
    "Successful prediction percentage: "
    f"{(target == target_predicted).mean()*100:.2f}%"
)
```
