```python
import pandas as pd
```

```python
adult_census = pd.read_csv("./dataset/adult-census-numeric.csv")
adult_census.head()
```

```python
target_class = "class"
target = adult_census[target_class]
target.head()
```

```python
data = adult_census.drop(columns=[target_class])
data.head()
```

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier()
_ = model.fit(data, target)
```

```python
target_predicted = model.predict(data)
```

```python
target_predicted
```

```python
print(
    "Number of correct prediction: "
    f"{(target == target_predicted).sum()} / {adult_census.shape[0]}"
)
print(
    "Successful prediction percentage: "
    f"{(target == target_predicted).mean()*100:.2f}%"
)
```

```python
adult_census_test = pd.read_csv("./dataset/adult-census-numeric-test.csv")
adult_census_test.head()
```

```python
print(
  f"Number of samples: {adult_census_test.shape[0]}"
)
```

```python
target_test = adult_census_test[target_class]
data_test = adult_census_test.drop(columns=[target_class])
```

```python
accuracy = model.score(data_test, target_test)
model_name = model.__class__.__name__

print(f"The test accuracy using a {model_name} is {accuracy:.3f}")
```

```python
test_target_predicted = model.predict(data_test)

print(
    "Number of correct prediction: "
    f"{(target_test == test_target_predicted).sum()} / {adult_census_test.shape[0]}"
)
print(
    "Successful prediction percentage: "
    f"{(target_test == test_target_predicted).mean()*100:.2f}%"
)
```
