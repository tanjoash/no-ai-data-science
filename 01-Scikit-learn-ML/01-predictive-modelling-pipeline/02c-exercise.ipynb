```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
```

```python
adult_census = pd.read_csv("./dataset/adult-census.csv")
```

```python
target_name = "class"
numerical_columns = ["age", "capital-gain", "capital-loss", "hours-per-week"]
target, data = adult_census[target_name], adult_census.drop(columns=target_name)
```

```python
data_numeric = data[numerical_columns]
```

```python
data_train, data_test, target_train, target_test = train_test_split(
  data_numeric, target, random_state=42, test_size=0.25
)
```

```python
dummy_model = DummyClassifier(strategy="constant", constant=" >50K")
dummy_model.fit(data_train, target_train)
dummy_model.score(data_test, target_test)
```

```python
dummy_model = DummyClassifier(strategy="constant", constant=" <=50K")
dummy_model.fit(data_train, target_train)
dummy_model.score(data_test, target_test)
```

Any Model that gives an accuracy lower than the later dummy_model would not be helpful. 0.766 is expected since 3/4 of the data is " <=50K".

```python
dummy_model = DummyClassifier(strategy="most_frequent")
dummy_model.fit(data_train, target_train)
dummy_model.score(data_test, target_test)
```
