```python
import pandas as pd
import seaborn as sns
```

```python
penguin = pd.read_csv("./dataset/penguins.csv")
penguin.head()
```

```python
studyName = penguin["Delta 15 N (o/oo)"].value_counts()
# Stage, Region are singular variables
studyName
```

```python
categorical_columns = [
  "studyName",
  "Region",
  "Island",
  #"Individual ID",
  "Clutch Completion",
  #"Date Egg",
  "Sex"
]
numerical_columns = [
  "Sample Number",
  "Culmen Length (mm)",
  "Culmen Depth (mm)",
  "Flipper Length (mm)",
  "Body Mass (g)",
  "Delta 15 N (o/oo)",
  "Delta 13 C (o/oo)"
]
open_columns = [
  "Comments"
]
```

```python
penguin_species = penguin["Species"].value_counts()
# Stage, Region are singular variables
penguin_species
```

```python
_ = penguin[numerical_columns].hist(figsize=(20,14))
```

```python
_ = sns.pairplot(
  data=penguin,
  hue="Species",
  vars=categorical_columns,
  plot_kws={"alpha": 0.2},
  height=3,
  diag_kind="hist",
  diag_kws={"bins": 30}
)
```

```python
_ = sns.scatterplot(
    x="Culmen Length (mm)",
    y="Culmen Depth (mm)",
    data=penguin,
    hue="Species",
    alpha=0.5,
)
```
