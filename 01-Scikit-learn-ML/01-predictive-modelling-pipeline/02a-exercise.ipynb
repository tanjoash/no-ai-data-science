```python
import pandas as pd

adult_census = pd.read_csv("./dataset/adult-census-numeric.csv")
data = adult_census.drop(columns="class")
target = adult_census["class"]
```

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=50)
_ = model.fit(data, target)
```

```python
number_of_data_points = 10
target_predicted = model.predict(data[:number_of_data_points])
```

```python
print(
  f"{(target[:number_of_data_points] == target_predicted).sum()} / {number_of_data_points}"
)
```

```python
target_predicted = model.predict(data)

print(
  f"{(target == target_predicted).sum()} / {adult_census.shape[0]}"
)
print(
  f"Accuracy: {(target == target_predicted).mean() * 100 :.2f} "
)
```

```python
accuracy = model.score(data, target)

print(f"Accuracy: {accuracy*100:.2f}")
```

```python
adult_census_test = pd.read_csv("./dataset/adult-census-numeric-test.csv")
test_data = adult_census_test.drop(columns="class")
test_target = adult_census_test["class"]

accruacy_test = model.score(test_data, test_target)
print(f"Accuracy: {accruacy_test*100:.2f}")
```
