```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
adult_census = pd.read_csv("./dataset/adult-census.csv")
target_name = "class"
target = adult_census[target_name]
data = adult_census.drop(columns=target_name)
```

```python
numerical_columns = ["age", "capital-gain", "capital-loss", "hours-per-week"]
data_numeric = data[numerical_columns]
```

```python
data_train, data_test, target_train, target_test = train_test_split(
    data_numeric, target, random_state=42
)
data_train.describe()
```

### Scaling
Models that rely on distance between samples should be trained on normalised features. Models such as logistic regression and k-nearest neighbors can benefit from normalization.

```python
scaler = StandardScaler()
scaler.fit(data_train)
data_train_scaled = scaler.transform(data_train) # Output: np arrray

## This can also be done in one line
data_train_scaled = scaler.fit_transform(data_train) # Output: np array

## By default, it would change into a np array but we can set property transform pandas to preserve the column name
scaler = StandardScaler().set_output(transform="pandas")
data_train_scaled = scaler.fit_transform(data_train) # Ouput: pd dataframe
```

```python
import time
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline

model = make_pipeline(StandardScaler(), LogisticRegression())
model
```

```python
model.named_steps
```

```python
start = time.time()
model.fit(data_train, target_train)
elapsed_time = time.time() - start
print(elapsed_time)
```

```python
predicted_target = model.predict(data_test)
predicted_target
```

```python
score = model.score(data_test, target_test)
model_name = model.__class__.__name__
print(
    f"The accuracy using a {model_name} is {score:.3f} "
    f"with a fitting time of {elapsed_time:.3f} seconds "
    f"in {model[-1].n_iter_[0]} iterations"
)
```

Comparing to the simple Regression without preprocessing, the transformed data is a little faster.
